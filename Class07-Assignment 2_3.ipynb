Q1. Division with Exception Handling

Write a program that takes two numbers and safely performs division. Handle division by zero using try-except.

```python
try:
    num1 = int(input("Please enter first number"))
    num2 = int(input("Plese enter second number"))
    division = num1/num2
    print(f"Division of numbers is: {division}")
except ZeroDivisionError:
    print("Number cannot be divide by Zero. Please enter correct number.")
except ValueError:
    print("Please enter integer value only")
```

Q2. String to Integer Conversion

Given a list:
data = ["10", "20", "abc", "30"]
Convert all values to integers. Skip invalid values using exception handling.

```python
data = ["10","20","abc","30"]
integer_data = []
for item in data:
    try:
        integer_data.append(int(item))
    except ValueError:
        print(f"Skipping invalid value: {item}")
print(integer_data)
```

Q3. Multiple Exception Handling

Write a program that:
Takes user input Converts it to integer Divides 100 by that number
Handle:
ValueError ZeroDivisionError

```python
try:
    number = input("Please enter a number to divide 100:")
    num1 = int(number)
    
    result = 100/num1
    print(f"100/{num1} = {result}")
except ValueError:
    print("Incorrect Input. Please enter digits only")
except ZeroDivisionError:
    print("Cannot divide by Zero.")
```

Q5. Using assert

Write a program that checks if a number is positive using assert.

```python
user_input = input("Please enter a number: ")
try:
    num = int(user_input)
    assert num > 0, "Number must be positive"
    
    print("Number is positive")
except ValueError:
    print("Invalid input. Please enter an integer.")
except AssertionError as e:
    print(f"Error: {e}")
```
